```python
import pandas as pd
import numpy as np

# Load your dataset
df = pd.read_csv("customer_shopping_behavior.csv")

# Number of new rows to generate
new_rows = 5000

# Generate synthetic rows
synthetic_df = df.sample(n=new_rows, replace=True).reset_index(drop=True)

# Create new Customer IDs
start_id = df["Customer ID"].max() + 1
synthetic_df["Customer ID"] = range(start_id, start_id + new_rows)

# Add some randomness
synthetic_df["Age"] = (
    synthetic_df["Age"] + np.random.randint(-5, 6, size=new_rows)
).clip(18, 70)

synthetic_df["Purchase Amount (USD)"] = (
    synthetic_df["Purchase Amount (USD)"] +
    np.random.randint(-10, 11, size=new_rows)
).clip(5, 500)

synthetic_df["Review Rating"] = (
    synthetic_df["Review Rating"] +
    np.random.uniform(-0.4, 0.4, size=new_rows)
).clip(1.0, 5.0).round(1)

# Combine original and new data
extended_df = pd.concat([df, synthetic_df], ignore_index=True)

# Save
extended_df.to_csv("customer_shopping_behavior_extended.csv", index=False)

print(extended_df.shape)
```
