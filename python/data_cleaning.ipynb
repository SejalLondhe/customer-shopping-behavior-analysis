```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior_extended.csv')
```

```python
df.describe(include= 'all')
```

```python
df.isnull().sum()
```

```python
# impute null value by category means taking median of rating by each category of clothes or footwear etc.
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#  convert columns names into sneak casing
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a columns age_group to group customer into group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create columns purchased_frequency_days (for this we will create a mapping dictionary to map the frequency of purchases to number of days)
print(df["frequency_of_purchases"].unique())
```

```python

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(5)
```

```python
# can we actually need previous columns with same data inside lets see
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# both columns are same so we donth neet both at same time ket delete one columns
df = df.drop('promo_code_used',axis = 1)
```
